# 02 - Data Type Handling

Dataset used: `hr_employee_attrition_raw.csv`. Same raw file as Notebook 1, so it still has string-encoded numbers, unit suffixes, and inconsistent categories — the exact conditions that make data type handling necessary.

```python
import pandas as pd
import numpy as np
pd.set_option('display.max_columns', None)
df = pd.read_csv('hr_employee_attrition_raw.csv')
print(df.shape)
df.head(3)
```

## 1. Identifying Data Types

`df.dtypes` shows what pandas *currently thinks* each column is — not what it *should* be.

```python
df.dtypes
```

Notice `MonthlyIncome` and `YearsAtCompany` are `object`/string even though they represent numbers, and `JoinDate` is a string even though it represents a date. This mismatch is common in raw data pulled from spreadsheets, forms, or exports.

```python
df.select_dtypes(include='number').columns.tolist(), df.select_dtypes(include='object').columns.tolist()
```

## 2. Numerical Data

Numerical data supports arithmetic: mean, sum, comparison, scaling. Two kinds:
- **Continuous** — `DistanceFromHomeKM`, `MonthlyIncome` (once fixed)
- **Discrete / count** — `Age`, `YearsAtCompany`, `PerformanceRating`

For a column to behave numerically in pandas, its dtype must be `int64` or `float64` — not `object` holding numeric-looking strings.

```python
df['Age'].dtype, df['DistanceFromHomeKM'].dtype, df['MonthlyIncome'].dtype
```

```python
try:
    df['MonthlyIncome'].mean()
except TypeError as e:
    print('Error:', e)
```

`MonthlyIncome` looks numeric when printed, but it's stored as text, so `.mean()` fails. This is why checking `dtype`, not just eyeballing values, matters.

## 3. Categorical Data

Categorical data represents a fixed, limited set of labels (`Department`, `JobRole`, `Education`, `Gender`). Storing these as pandas' `category` dtype instead of generic `object`:
- Uses less memory (labels stored as integer codes internally)
- Signals intent to downstream code and plotting libraries
- Enables ordered comparisons when categories have a natural order (e.g. Education level)

```python
df['Department'].nunique(), df['Department'].unique()
```

```python
df['Department'] = df['Department'].astype('category')
df['Department'].dtype
```

Education has a natural order, so it can be made an **ordered** category:

```python
education_order = ['High School', "Bachelor's", "Master's", 'PhD']
df['Education'] = pd.Categorical(df['Education'], categories=education_order, ordered=True)
df['Education'].dtype
```

```python
# Ordered categories support comparison operators
df[df['Education'] > 'High School']['Education'].value_counts()
```

## 4. Boolean Data

Boolean columns hold only two states (True/False). `OverTime` and `Attrition` are logically boolean but arrive as inconsistent Yes/No strings.

```python
df['OverTime'].unique()
```

```python
overtime_clean = df['OverTime'].str.strip().str.lower().map({'yes': True, 'y': True, 'no': False, 'n': False})
overtime_clean.dtype, overtime_clean.value_counts(dropna=False)
```

Mapping to actual booleans (rather than leaving 'Yes'/'yes'/'Y' as separate strings) makes filtering reliable: `df[overtime_clean]` instead of juggling six spelling variants.

## 5. Date/Time Data

`JoinDate` is stored as text (`'2019-07-23'`). As a string it can't be used for calculations like tenure, sorting by recency, or extracting the join year.

```python
df['JoinDate'].dtype
```

```python
join_date = pd.to_datetime(df['JoinDate'])
join_date.dtype
```

```python
# Now date arithmetic works
tenure_days = (pd.Timestamp('2025-01-01') - join_date).dt.days
tenure_days.describe()
```

```python
join_date.dt.year.value_counts().sort_index()
```

## 6. String Data

`EmployeeID`, `JobRole`, and free-text-like fields stay as strings (`object` dtype in pandas). String dtype is appropriate when values are identifiers or labels that won't be aggregated numerically. Pandas' `.str` accessor gives vectorized string operations.

```python
df['EmployeeID'].dtype
```

```python
df['EmployeeID'].str.startswith('EMP').all(), df['EmployeeID'].str.len().unique()
```

## 7. Converting Data Types

### `astype()`
Best for **clean, already-valid** values — straightforward casts like int to float, or object to category. It has no built-in handling for messy/invalid values: a single unparseable value raises an error.

```python
# Integer -> Float
df['PerformanceRating'].astype(float).dtype
```

```python
# astype() fails hard on invalid values - it does not silently coerce
try:
    df['MonthlyIncome'].astype(float)
except ValueError as e:
    print('astype() failed:', str(e)[:120])
```

### `to_numeric()`
Purpose-built for converting to numeric types, with an `errors=` parameter to control what happens when a value can't be converted:
- `errors='raise'` (default) — stop on first bad value
- `errors='coerce'` — invalid values become `NaN`
- `errors='ignore'` — leave the column unchanged (deprecated in modern pandas; prefer try/except)

```python
# String -> Numeric, after stripping unit suffixes
income_clean = df['MonthlyIncome'].str.replace(' USD', '', regex=False)
income_numeric = pd.to_numeric(income_clean, errors='coerce')
print(income_numeric.dtype)
print('New NaNs introduced by bad values:', income_numeric.isna().sum() - df['MonthlyIncome'].isna().sum())
```

```python
# Inspect what actually failed to convert
still_bad = df.loc[income_numeric.isna() & df['MonthlyIncome'].notna(), 'MonthlyIncome']
still_bad.head(10)
```

### `to_datetime()`
Converts strings to `datetime64` with the same `errors=` control, plus a `format=` argument to speed up and validate parsing against an expected pattern.

```python
# String -> Date
join_date_coerced = pd.to_datetime(df['JoinDate'], format='%Y-%m-%d', errors='coerce')
join_date_coerced.isna().sum()
```

### Numeric -> Categorical
Binning a continuous number into labeled groups (`pd.cut`) converts it to a category — useful when a model or report needs bands rather than raw values.

```python
# Numeric -> Categorical
age_band = pd.cut(df['Age'].clip(upper=75), bins=[0, 25, 35, 45, 100], labels=['<=25', '26-35', '36-45', '46+'])
age_band.dtype, age_band.value_counts()
```

## 8. Handling Invalid Conversions

Real columns rarely convert cleanly on the first try. The safe pattern is:
1. Attempt conversion with `errors='coerce'`
2. Compare the resulting `NaN` count to the *original* `NaN` count
3. Investigate the difference before deciding to drop, fix, or impute

```python
def coerce_report(original: pd.Series, converted: pd.Series, name: str):
    new_nans = converted.isna().sum() - original.isna().sum()
    print(f'{name}: {new_nans} value(s) failed to convert and became NaN')
    if new_nans > 0:
        print(original[converted.isna() & original.notna()].unique()[:10])

coerce_report(df['MonthlyIncome'], income_numeric, 'MonthlyIncome')
```

**Never convert with `errors='coerce'` and move on without checking what was lost.** Silently turning bad values into `NaN` can hide data entry errors (e.g. `'41019.93508181599'` — a plausible float but a suspicious outlier) or genuine formatting issues (e.g. `'5208.92 USD'`) that need separate handling, not blind imputation.

## 9. Detecting Incorrect Data Types

Three practical checks for spotting columns that have the wrong dtype:

```python
# Check 1: object columns that are actually fully numeric
for col in df.select_dtypes(include='object').columns:
    converted = pd.to_numeric(df[col], errors='coerce')
    pct_numeric = converted.notna().mean() * 100
    if pct_numeric > 80:
        print(f"{col}: {pct_numeric:.1f}% of values parse as numeric but dtype is {df[col].dtype}")
```

```python
# Check 2: low-cardinality object columns that should be categorical
for col in df.select_dtypes(include='object').columns:
    n_unique = df[col].nunique()
    if n_unique < 15:
        print(f'{col}: only {n_unique} unique values -> candidate for category dtype')
```

```python
# Check 3: object columns that actually hold dates
for col in df.select_dtypes(include='object').columns:
    parsed = pd.to_datetime(df[col], errors='coerce')
    if parsed.notna().mean() > 0.9:
        print(f'{col}: {parsed.notna().mean()*100:.1f}% parse as dates -> candidate for datetime64')
```

## 10. Why Correct Data Types Matter

- **Correctness of computation** — `.mean()`, `.sum()`, and comparisons on numeric-looking strings either fail outright or produce wrong results (string sort order, not numeric order).
- **Memory efficiency** — `category` dtype and smaller numeric types (e.g. `int32` vs `int64`) cut memory usage substantially on large datasets.
- **Model compatibility** — machine learning algorithms expect numeric input; dates and booleans need to be encoded, and that encoding step assumes the column is already the correct semantic type.
- **Correct sorting and grouping** — a numeric column stored as string sorts `'10' < '2'` (alphabetically), which silently breaks ranking and grouping logic.
- **Catching data quality problems early** — attempting a type conversion is often the fastest way to *discover* messy values (unit suffixes, stray text, wrong formats) that would otherwise pass silently into modeling.
