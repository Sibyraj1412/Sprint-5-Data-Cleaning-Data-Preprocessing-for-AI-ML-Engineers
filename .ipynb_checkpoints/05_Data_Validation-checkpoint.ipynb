# Notebook 5 – Data Validation
## What is Data Validation?

Cleaning removes missing values and duplicates. **Validation** goes further -
it checks whether the values that *are* there actually make sense.

A row can be "complete" (no nulls, no duplicates) and still be wrong:
an age of `999`, a salary of `-500`, or an employee who "joined" the company
before they were born. Validation is the set of rules that catches these.

Let's load the data and validate it, rule by rule.

```python
import pandas as pd

df = pd.read_csv("hr_employee_attrition_raw.csv")
df.shape
```

## 1. Range Validation
**Rule: Age cannot be negative, and must fall within a believable human range.**

Real people don't work at age `999`. Let's check.

```python
# Range Validation — Age
invalid_age = df[(df["Age"] < 18) | (df["Age"] > 75)]
print("Invalid age rows:", len(invalid_age))
invalid_age[["EmployeeID", "Age"]]
```

Sure enough - some rows have Age = `150`, `200`, even `999`. These are almost
certainly data entry glitches (extra digits, typos), not real ages. A range
rule catches what a null-check never would, since these values aren't missing - they're just wrong.

## 2. Type Validation
**Rule: Salary (MonthlyIncome) must be a number, not text.**

`MonthlyIncome` should be numeric, but some entries have extra text like
`"4210.69 USD"` mixed in, which forces the whole column to be stored as text.

```python
# Type Validation — MonthlyIncome should be numeric
non_numeric = df[~df["MonthlyIncome"].astype(str).str.replace(",", "").str.match(r"^-?\d+\.?\d*$")]
print("Non-numeric income entries:", len(non_numeric))
non_numeric[["EmployeeID", "MonthlyIncome"]].head()
```

## 3. Format Validation
**Rule: MonthlyIncome must not contain currency text like "USD".**

Format validation checks the *shape* of a value, not just its type.
Once we know which rows have text, we can clean them consistently.

```python
# Format Validation — strip and standardize
df["MonthlyIncome_clean"] = (
    df["MonthlyIncome"].astype(str).str.replace(" USD", "", regex=False).str.strip()
)
df["MonthlyIncome_clean"] = pd.to_numeric(df["MonthlyIncome_clean"], errors="coerce")

# Salary cannot be negative
negative_salary = df[df["MonthlyIncome_clean"] < 0]
print("Negative salary rows:", len(negative_salary))
```

## 4. Category Validation
**Rule: Gender and Department must belong to an accepted, standardized set.**

Right now `Gender` has values like `'Female'`, `'FEMALE'`, `'male'`, `'M'`,
`'F'` - all meaning the same thing but written differently. Same story for
`Department`, with `'sales'`, `'Sales'`, `'SALES'`, `'Sales '` (trailing space).

```python
# Category Validation — check for inconsistent labels
print("Raw Gender values:", df["Gender"].unique())
print("Raw Department values:", df["Department"].unique())

# Standardize
gender_map = {"m": "Male", "male": "Male", "f": "Female", "female": "Female"}
df["Gender_clean"] = df["Gender"].str.strip().str.lower().map(gender_map)

df["Department_clean"] = df["Department"].str.strip().str.title()

accepted_departments = {"Finance", "Sales", "Hr", "Support", "Engineering", "Marketing"}
invalid_dept = df[~df["Department_clean"].isin(accepted_departments)]
print("Rows with unexpected department after cleanup:", len(invalid_dept))
```

## 5. Null Validation
**Rule: Critical fields should not be missing.**

Not every missing value is equally serious. A missing `Education` is minor;
a missing `Gender` or `MonthlyIncome` might matter more depending on your
analysis. Null validation means deciding - per column - what's acceptable.

```python
# Null Validation — check missing counts per column
df.isna().sum().sort_values(ascending=False)
```

## 6. Business Rule Validation
**Rule: Percentage-style fields must fall in their defined business range.**

`JobSatisfaction` and `PerformanceRating` aren't free numbers - they're
ratings meant to fall within a fixed scale (e.g., 1 to 4 in this dataset).
A business rule enforces that meaning, not just "is it a number."

```python
# Business Rule Validation — ratings must be within their defined scale
invalid_satisfaction = df[~df["JobSatisfaction"].between(1, 4) & df["JobSatisfaction"].notna()]
invalid_rating = df[~df["PerformanceRating"].between(1, 4)]

print("Invalid JobSatisfaction rows:", len(invalid_satisfaction))
print("Invalid PerformanceRating rows:", len(invalid_rating))
```

## 7. Referential Validation
**Rule: Every EmployeeID should be unique and internally consistent.**

Referential validation checks that a record correctly "refers" to something
that should be unique or linked - like an ID that shouldn't repeat across
otherwise-different people.

```python
# Referential Validation — EmployeeID should not repeat with conflicting data
id_counts = df["EmployeeID"].value_counts()
repeated_ids = id_counts[id_counts > 1]
print("EmployeeIDs appearing more than once:", len(repeated_ids))
```

## 8. Date Validation
**Rule: Date of joining cannot come before the employee was even old enough
to work.**

We don't have a direct date-of-birth column, but we can estimate it from
`Age` and `JoinDate`, then check whether the person would've joined the
company at an impossibly young age (e.g., under 15).

```python
# Date Validation
df["JoinDate"] = pd.to_datetime(df["JoinDate"], errors="coerce")

# Estimate age at joining using current age minus years passed since joining
years_since_join = (pd.Timestamp("2026-09-01") - df["JoinDate"]).dt.days / 365.25
age_at_joining = df["Age"] - years_since_join

impossible_join = df[age_at_joining < 15]
print("Rows where employee 'joined' before a believable working age:", len(impossible_join))
impossible_join[["EmployeeID", "Age", "JoinDate"]].head()
```

## 9. Constraint Validation
**Rule: Fixed/constant columns must actually stay constant.**

`EmployeeCountFlag` looks like it's meant to always equal `1` (a constant
system flag). A constraint check confirms whether that assumption actually
holds - if it doesn't, something upstream may be broken.

```python
# Constraint Validation - this column should always be exactly 1
constraint_violations = df[df["EmployeeCountFlag"] != 1]
print("Rows breaking the constant-flag constraint:", len(constraint_violations))
```

## Putting It All Together - A Simple Validation Function

Rather than running checks one by one, we can wrap them into a single
reusable function that reports every problem at once.

```python
def validate_employee_data(df):
    report = {}

    report["invalid_age"] = df[(df["Age"] < 18) | (df["Age"] > 75)].shape[0]
    report["negative_or_missing_salary"] = df["MonthlyIncome_clean"].lt(0).sum() if "MonthlyIncome_clean" in df else None
    report["invalid_job_satisfaction"] = df[~df["JobSatisfaction"].between(1, 4) & df["JobSatisfaction"].notna()].shape[0]
    report["invalid_performance_rating"] = df[~df["PerformanceRating"].between(1, 4)].shape[0]
    report["duplicate_employee_ids"] = df["EmployeeID"].duplicated().sum()
    report["broken_flag_constraint"] = df[df["EmployeeCountFlag"] != 1].shape[0]

    return pd.Series(report)

validate_employee_data(df)
```
